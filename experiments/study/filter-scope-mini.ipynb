# Mini filter-scope comparison

Editable report generator. Run All reads the hash-bound bundle and writes `filter-scope-mini.html`; it never trains, plays games, or contacts trackers. Actor-only is an additional MTG hypothesis, not an exact Ataraxos prescription. Every seed and original endpoint is retained.

```python
from pathlib import Path
import json
import sys
ROOT = Path.cwd()
if not (ROOT / 'manabot').is_dir():
    ROOT = ROOT.parents[1] if ROOT.name == 'study' else ROOT
sys.path.insert(0, str(ROOT))
from experiments.runners.filter_scope_report import load_bundle, evidence, summarize, summary_html, monitoring_figures, final_study
from manabot.training.experiment_report import diagnostic_figures, write_dashboard
from manabot.training.report_study import study_strength_figures
DATA = ROOT / 'experiments/data/etu105/filter-scope-mini.json.gz'
OUTPUT = ROOT / 'experiments/study/filter-scope-mini.html'
bundle = load_bundle(DATA)
saved = evidence(bundle, DATA)
summary = summarize(bundle)
study = final_study(bundle)
```

## Interpretation

The final held-out greedy comparison is the endpoint result. Random is a separately frozen diagnostic. Monitoring deals were inspected repeatedly; their curves and common-cost comparisons are exploratory. Four seat/deck legs remain together in all uncertainty calculations; training seeds are the experimental unit. The same architecture, quantile, magnitude floor, schedules, opponents and world are held fixed. Changing critic coverage also changes the shared policy representation and subsequent actor selection.

```python
# Editable plot selection. Per-update diagnostics retain their original clocks.
sections = [('All observed monitoring milestones', 'comparisons', monitoring_figures(bundle)[:2])]
# Optional deeper plots: diagnostic_figures(saved), study_strength_figures(study).
effect = summary['effects']['final-greedy']['actor_minus_actor_critic']
NOTES = (f"Actor-only remains unpromoted: held-out greedy effect {effect['mean']*100:+.2f} "
         f"[{effect['lower']*100:+.2f}, {effect['upper']*100:+.2f}] percentage points, "
         f"at {summary['actor_cost_ratio']:.2f}× mean training wall cost. "
         "Three paired seeds, final raw checkpoints at 1,240 updates / 317,440 transitions. "
         "No demonstrated equal-cost payoff; not an equivalence claim or a verdict for larger budgets. "
         "ETU-118 recipe selection remains independent. S3 publication is blocked by credentials; "
         "originals and 48 hash-verified local exports remain retained. W&B has the metric projections.")
report = write_dashboard(saved, OUTPUT, question='Does training the critic on every row justify its extra cost?',
    docs='../../docs/experiment-metrics.md', sections=sections, notes=NOTES, study=study)
html = report.read_text()
html = html.replace('<section>', summary_html(bundle) + '<section>', 1)
assert 'Every seed, without selection' in html
raw = OUTPUT.with_name(OUTPUT.stem + '-diagnostics.json')
html = html.replace(raw.name, '../data/etu105/filter-scope-mini.json.gz').replace('Download all raw scalar diagnostics (JSON)', 'Download compact evidence and all scalar diagnostics (JSON.gz)')
raw.unlink()
report.write_text(html)
(ROOT / 'experiments/data/etu105/analysis.json').write_text(json.dumps(summary, indent=2, sort_keys=True) + '\n')
print(f'Read-only report: {OUTPUT}')
```

Lower value loss across different selected sample populations cannot establish better value calibration. No exploitability, S1–S5 competence, human-play or chapter success is inferred.
